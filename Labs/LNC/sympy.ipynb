```python
import sys
sys.set_int_max_str_digits(0)  # 무제한 출력 허용 (필수!)

def karatsuba(x, y):
    """개선된 Karatsuba 곱셈: Python 내장 int 사용 + 안전한 pow 계산 + threshold 높임"""
    if x == 0 or y == 0:
        return 0
    # 작은 수는 Python 기본 곱셈으로 (가장 빠름)
    if len(str(x)) < 100 or len(str(y)) < 100:  # threshold를 충분히 높여서 overflow 줄임
        return x * y
    
    m = max(len(str(x)), len(str(y))) // 2
    pow_m = 10 ** m  # 여기서 m이 너무 크면 overflow 가능성 있지만, 재귀 깊이 제한으로 막음
    
    a, b = divmod(x, pow_m)
    c, d = divmod(y, pow_m)
    
    z0 = karatsuba(b, d)
    z2 = karatsuba(a, c)
    z1 = karatsuba(a + b, c + d)  # 재귀 순서 바꿔서 z2 먼저 계산 (안정성 ↑)
    
    pow_2m = pow_m ** 2  # 10**(2*m) 대신 이렇게 계산 (overflow 덜 남)
    return z2 * pow_2m + (z1 - z2 - z0) * pow_m + z0

def karatsuba_factorial(n):
    """Product Tree + Karatsuba를 사용한 팩토리얼 계산"""
    if n <= 1:
        return 1
    
    def product_tree(start, end):
        if start > end:
            return 1
        if start == end:
            return start  # leaf: 그냥 숫자 반환
        if end - start + 1 <= 20:  # 작은 범위는 기본 곱셈으로 빠르게
            res = 1
            for i in range(start, end + 1):
                res *= i
            return res
        
        mid = (start + end) // 2
        left = product_tree(start, mid)
        right = product_tree(mid + 1, end)
        return karatsuba(left, right)  # 큰 곱셈에만 Karatsuba 적용
    
    return product_tree(1, n)
```

```python
print(karatsuba_factorial(1_000_000))
```
