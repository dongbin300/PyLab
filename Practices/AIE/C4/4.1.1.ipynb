```python
# These are the libraries will be used for this lab.
import numpy as np 
import pandas as pd

import matplotlib.pyplot as plt
%matplotlib inline

import torch
```

```python
# Plot vecotrs, please keep the parameters in the same length
# @param: Vectors = [{"vector": vector variable, "name": name of vector, "color": color of the vector on diagram}]
    
def plotVec(vectors):
    ax = plt.axes()
    
    # For loop to draw the vectors
    for vec in vectors:
        ax.arrow(0, 0, *vec["vector"], head_width = 0.05,color = vec["color"], head_length = 0.1)
        plt.text(*(vec["vector"] + 0.1), vec["name"])
    
    plt.ylim(-2,2)
    plt.xlim(-2,2)
```

```python
# int 변수들을 1차원 텐서로 변환
# type(): 텐서의 타입
# dtype: 텐서 안의 멤버들의 타입

ints_to_tensor = torch.tensor([0, 1, 2, 3, 4])
print("The dtype of tensor object after converting it to tensor: ", ints_to_tensor.dtype)
print("The type of tensor object after converting it to tensor: ", ints_to_tensor.type())
```

```python
# type(__TENSOR__): torch.Tensor
type(ints_to_tensor)
```

```python
# float 변수들을 1차원 텐서로 변환
# type(): 텐서의 타입
# dtype: 텐서 안의 멤버들의 타입

floats_to_tensor = torch.tensor([0.0, 1.0, 2.0, 3.0, 4.0])
print("The dtype of tensor object after convertiang it to tensor: ", floats_to_tensor.dtype)
print("The type of tensor object after converting it to tensor: ", floats_to_tensor.type())
```

```python
# 텐서로 변환할 때 데이터타입을 명시
# 데이터타입을 명시하면 텐서타입도 자동으로 따라감

list_floats=[0.0, 1.0, 2.0, 3.0, 4.0]

floats_int_tensor=torch.tensor(list_floats,dtype=torch.int64)

print("The dtype of tensor object is: ", floats_int_tensor.dtype)
print("The type of tensor object is: ", floats_int_tensor.type())
```

```python
# tensor대신 FloatTensor로 텐서타입 명시
new_float_tensor = torch.FloatTensor([0, 1, 2, 3, 4])
new_float_tensor.type()
print("The type of the new_float_tensor:", new_float_tensor.type())
```

```python
# int텐서를 float텐서로 변환
old_int_tensor = torch.tensor([0, 1, 2, 3, 4])
new_float_tensor = old_int_tensor.type(torch.FloatTensor)
print("The type of the new_float_tensor:", new_float_tensor.type())
```

```python
# size(): 사이즈 반환, 1차원이면 [n] 형태로 반환
# ndimension(): 차원 반환
print("The size of the new_float_tensor: ", new_float_tensor.size())
print("The dimension of the new_float_tensor: ",new_float_tensor.ndimension())
```

```python
# 2차원 텐서(5*1)로 변환
# size(): 2차원이면 [m, n] 형태로 반환
twoD_float_tensor = new_float_tensor.view(5, 1)
print("Original Size: ", new_float_tensor.size())
print("Size after view method", twoD_float_tensor.size())
```

```python
# 2차원 텐서로 변환
# -1을 넣으면 전체 사이즈
twoD_float_tensor = new_float_tensor.view(-1, 1)
print("Original Size: ", new_float_tensor.size())
print("Size after view method", twoD_float_tensor.size())
```

```python
# numpy배열을 텐서배열로 변환
# numpy float는 float64(double)라서 텐서배열로 변환시 DoubleTensor타입으로 변환됨
numpy_array = np.array([0.0, 1.0, 2.0, 3.0, 4.0])
new_tensor = torch.from_numpy(numpy_array)

print("The dtype of new tensor: ", new_tensor.dtype)
print("The type of new tensor: ", new_tensor.type())
```

```python
# 텐서배열을 다시 numpy배열로 변환
back_to_numpy = new_tensor.numpy()
print("The numpy array from tensor: ", back_to_numpy)
print("The dtype of numpy array: ", back_to_numpy.dtype)
```

```python
# from_numpy(), numpy()로 복사한 값들은 원본이 업데이트가 되면 자동으로 업데이트가 됨
numpy_array[:] = 0
print("The new tensor points to numpy_array : ", new_tensor)
print("and back to numpy array points to the tensor: ", back_to_numpy)
```

```python
# panda배열을 텐서배열로 변환
# panda float는 float64(double)라서 텐서배열로 변환시 DoubleTensor타입으로 변환됨
pandas_series=pd.Series([0.1, 2, 0.3, 10.1])
new_tensor=torch.from_numpy(pandas_series.values)
print("The new tensor from numpy array: ", new_tensor)
print("The dtype of new tensor: ", new_tensor.dtype)
print("The type of new tensor: ", new_tensor.type())
```

```python
# __TENSOR__[n]: tensor(n) 형태로 값을 반환
# __TENSOR__[n].item(): 값을 반환
this_tensor=torch.tensor([0,1, 2,3]) 

print("the first item is given by",this_tensor[0].item(),"the first tensor value is given by ",this_tensor[0])
print("the second item is given by",this_tensor[1].item(),"the second tensor value is given by ",this_tensor[1])
print("the third  item is given by",this_tensor[2].item(),"the third tensor value is given by ",this_tensor[2])
```

```python
# tolist(): 텐서배열을 단순배열로 변환
torch_to_list=this_tensor.tolist()

print('tensor:', this_tensor,"\nlist:",torch_to_list)
```

```python
# 텐서배열 값 변경
tensor_sample = torch.tensor([20, 1, 2, 3, 4])
print("Inital value on index 0:", tensor_sample[0])
tensor_sample[0] = 100
print("Modified tensor:", tensor_sample)
```

```python
# 텐서배열 값 변경
print("Inital value on index 4:", tensor_sample[4])
tensor_sample[4] = 0
print("Modified tensor:", tensor_sample)
```

```python
# 텐서배열 slice
subset_tensor_sample = tensor_sample[1:4]
print("Original tensor sample: ", tensor_sample)
print("The subset of tensor sample:", subset_tensor_sample)
```

```python
# 텐서배열 값 다중 할당
print("Inital value on index 3 and index 4:", tensor_sample[3:5])
tensor_sample[3:5] = torch.tensor([300.0, 400.0])
print("Modified tensor:", tensor_sample)
```

```python
# ,로 다중 인덱스 선택
selected_indexes = [3, 4]
subset_tensor_sample = tensor_sample[selected_indexes]
print("The inital tensor_sample", tensor_sample)
print("The subset of tensor_sample with the values on index 3 and 4: ", subset_tensor_sample)
```

```python
# 인덱스1, 인덱스3만 선택해서 100000으로 값 변경
print("The inital tensor_sample", tensor_sample)
selected_indexes = [1, 3]
tensor_sample[selected_indexes] = 100000
print("Modified tensor with one value: ", tensor_sample)
```

```python
# 텐서배열의 평균, 표준편차
math_tensor = torch.tensor([1.0, -1.0, 1, -1])
mean = math_tensor.mean()
print("The mean of math_tensor: ", mean)
standard_deviation = math_tensor.std()
print("The standard deviation of math_tensor: ", standard_deviation)
```

```python
# 텐서배열의 최대, 최소값
max_min_tensor = torch.tensor([1, 1, 3, 5, 5])
max_val = max_min_tensor.max()
print("Maximum number in the tensor: ", max_val)
min_val = max_min_tensor.min()
print("Minimum number in the tensor: ", min_val)
```

```python
# 텐서배열의 sin 값
pi_tensor = torch.tensor([0, np.pi/2, np.pi])
sin = torch.sin(pi_tensor)
print("The sin result of pi_tensor: ", sin)
```

```python
# -2부터 2까지 5단계로 설정
len_5_tensor = torch.linspace(-2, 2, steps = 5)
print ("First Try on linspace", len_5_tensor)
```

```python
# -2부터 2까지 9단계로 설정
len_9_tensor = torch.linspace(-2, 2, steps = 9)
print ("Second Try on linspace", len_9_tensor)
```

```python
# 0부터 2파이 까지 100단계로 설정하고 sin값을 구해서 그래프로 표현

pi_tensor = torch.linspace(0, 2*np.pi, 100)
sin_result = torch.sin(pi_tensor)
plt.plot(pi_tensor.numpy(), sin_result.numpy())
```

```python
# 텐서 덧셈

u = torch.tensor([1, 0])
v = torch.tensor([0, 1])

w = u + v
print("The result tensor: ", w)
```

```python
plotVec([
    {"vector": u.numpy(), "name": 'u', "color": 'r'},
    {"vector": v.numpy(), "name": 'v', "color": 'b'},
    {"vector": w.numpy(), "name": 'w', "color": 'g'}
])
```
