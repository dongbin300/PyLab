1차원 선형 회귀: 예측

```python
import torch
```

```python
# w = 2, b = -1

w = torch.tensor(2.0, requires_grad = True)
b = torch.tensor(-1.0, requires_grad = True)
```

```python
# y = wx + b
# y = 2x - 1

def forward(x):
    yhat = w * x + b
    return yhat
```

```python
# x = 1
# y = 2x - 1 에 대입하면 1 나옴

x = torch.tensor([[1.0]])
yhat = forward(x)
print("The prediction: ", yhat)
```

```python
# x = [1, 2]

x = torch.tensor([[1.0], [2.0]])
print("The shape of x: ", x.shape)
```

```python
# y = 2x - 1 에 대입하면 [1, 3] 나옴

yhat = forward(x)
print("The prediction: ", yhat)
```

```python
# Import Class Linear
# Set random seed

from torch.nn import Linear
torch.manual_seed(1)
```

```python
# lr 1차원 선형 회귀 모델 생성(in = 1, out = 1)
# y = 0.5153x - 0.4414

lr = Linear(in_features=1, out_features=1, bias=True)
print("Parameters w and b: ", list(lr.parameters()))
```

```python
# 딕셔너리 상태 반환

print("Python dictionary: ",lr.state_dict())
print("keys: ",lr.state_dict().keys())
print("values: ",lr.state_dict().values())
```

```python
print("weight:",lr.weight)
print("bias:",lr.bias)
```

```python
# x = 1
# y = 0.5153 - 0.4414 = 0.0739

x = torch.tensor([[1.0]])
yhat = lr(x)
print("The prediction: ", yhat)
```

```python
# x = [1, 2]
# y = [0.0739, 0.5891]

x = torch.tensor([[1.0], [2.0]])
yhat = lr(x)
print("The prediction: ", yhat)
```

```python
# Library for this section

from torch import nn
```

```python
# 커스텀 선형회귀 클래스 구현

class LR(nn.Module):
    
    # Constructor
    def __init__(self, input_size, output_size):
        
        # Inherit from parent
        super(LR, self).__init__()
        self.linear = nn.Linear(input_size, output_size)
    
    # Prediction function
    def forward(self, x):
        out = self.linear(x)
        return out
```

```python
# in = 1, out = 1인 1차원 선형 회귀 클래스 인스턴스 생성
# y = -0.1939x + 0.4694

lr = LR(1, 1)
print("The parameters: ", list(lr.parameters()))
print("Linear model: ", lr.linear)
```

```python
# x = 1
# y = -0.1939 + 0.4694 = 0.2755

x = torch.tensor([[1.0]])
yhat = lr(x)
print("The prediction: ", yhat)
```

```python
# x = [1, 2]
# y = [0.2755, 0.0816]

x = torch.tensor([[1.0], [2.0]])
yhat = lr(x)
print("The prediction: ", yhat)
```

```python
print("Python dictionary: ", lr.state_dict())
print("keys: ",lr.state_dict().keys())
print("values: ",lr.state_dict().values())
```
