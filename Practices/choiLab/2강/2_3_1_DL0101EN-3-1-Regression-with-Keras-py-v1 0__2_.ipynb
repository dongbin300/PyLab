## Download and Clean the Data Set


Let's download the data and read it into a <em>pandas</em> dataframe.


```python
filepath='https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/DL0101EN/labs/data/concrete_data.csv'
concrete_data = pd.read_csv(filepath)

concrete_data.head()
```

So the first concrete sample has 540 cubic meter of cement, 0 cubic meter of blast furnace slag, 0 cubic meter of fly ash, 162 cubic meter of water, 2.5 cubic meter of superplaticizer, 1040 cubic meter of coarse aggregate, 676 cubic meter of fine aggregate. Such a concrete mix which is 28 days old, has a compressive strength of 79.99 MPa. 

첫 번째 콘크리트 시료는 시멘트 540 세제곱미터, 고로 슬래그 0 세제곱미터, 플라이애시 0 세제곱미터, 물 162 세제곱미터, 경화제 2.5 세제곱미터, 굵은 골재 1040 세제곱미터, 잔골재 676 세제곱미터로 구성되어 있습니다. 이 콘크리트는 28일 경과 후 압축강도가 79.99 MPa입니다.

#### Let's check how many data points we have
데이터 포인트가 몇 개인지 확인해 봅시다.

```python
concrete_data.shape
```

So, there are approximately 1000 samples to train our model on. Because of the few samples, we have to be careful not to overfit the training data.

따라서 모델 학습에 사용할 샘플은 약 1000개입니다. 샘플 수가 적기 때문에 학습 데이터에 과적합되지 않도록 주의해야 합니다.


Let's check the dataset for any missing values.
데이터셋에 누락된 값이 있는지 확인해 보겠습니다.

```python
concrete_data.describe()
```

```python
#비어있는 값 확인
concrete_data.isnull().sum()
```

데이터가 매우 깔끔하고 모델 구축에 바로 사용할 수 있을 것 같습니다.

#### Split data into predictors and target


The target variable in this problem is the concrete sample strength. Therefore, our predictors will be all the other columns.

이 문제에서 목표 변수는 콘크리트 시료 강도입니다. 따라서 예측 변수는 나머지 모든 열이 될 것입니다.

```python
concrete_data_columns = concrete_data.columns
```

```python
# predictors 강도 제외 하여 등록,  target 강도 값 만 (정답지)
predictors = concrete_data[concrete_data_columns[concrete_data_columns != 'Strength']] # all columns except Strength
target = concrete_data['Strength'] # Strength column
```

<a id="item2"></a>


Let's do a quick sanity check of the predictors and the target dataframes.
예측 변수와 목표 데이터프레임이 제대로 설정되었는지 간단히 확인해 보겠습니다

```python
predictors.head()
```

```python
target.head()
```

Finally, the last step is to normalize the data by substracting the mean and dividing by the standard deviation. 

마지막으로, 평균을 빼고 표준편차로 나누어 데이터를 정규화하는 것이 마지막 단계입니다.

```python
# 정규화 수싱 
# predictors - predictors.mean(): 각 데이터에서 해당 컬럼의 평균을 뺍니다. 이렇게 하면 데이터의 중심이 $0$이 됩니다.
# / predictors.std(): 그 결과를 표준편차로 나눕니다. 이렇게 하면 데이터가 퍼져 있는 정도(분산)가 일정하게 맞춰집니다.
predictors_norm = (predictors - predictors.mean()) / predictors.std()
predictors_norm.head()
```

Let's save the number of predictors to *n_cols* since we will need this number when building our network.


```python
n_cols = predictors_norm.shape[1] # number of predictors
```

<a id="item1"></a>


##  Import Keras Packages

##### Let's import the rest of the packages from the Keras library that we will need to build our regression model.


## Keras 패키지 가져오기

##### 회귀 모델을 구축하는 데 필요한 나머지 Keras 패키지를 가져오겠습니다.

```python
from keras.models import Sequential
from keras.layers import Dense
from keras.layers import Input
```

## Build a Neural Network


Let's define a function that defines our regression model for us so that we can conveniently call it to create our model.

회귀 모델을 정의하는 함수를 만들어, 필요할 때 편리하게 호출하여 모델을 생성할 수 있도록 해봅시다.

```python
# define regression model
def regression_model():
    # create model
    model = Sequential()
    model.add(Input(shape=(n_cols,)))
    model.add(Dense(50, activation='relu'))
    model.add(Dense(50, activation='relu'))
    model.add(Dense(1))
    
    # compile model
    model.compile(optimizer='adam', loss='mean_squared_error')
    return model
```

The above function create a model that has two hidden layers, each of 50 hidden units.

위 함수는 각각 50개의 은닉 유닛을 가진 두 개의 은닉층을 포함하는 모델을 생성합니다.

## Train and Test the Network


Let's call the function now to create our model.

이제 함수를 호출하여 모델을 생성해 보겠습니다.

```python
# build the model
model = regression_model()
```

Next, we will train and test the model at the same time using the *fit* method. We will leave out 30% of the data for validation and we will train the model for 100 epochs.

다음으로, `fit` 메서드를 사용하여 모델을 동시에 학습하고 테스트하겠습니다. 데이터의 30%는 검증용으로 남겨두고, 100 에포크 동안 모델을 학습시킬 것입니다.

```python
# fit the model 학습시작
model.fit(predictors_norm, target, validation_split=0.3, epochs=100, verbose=2)
```
