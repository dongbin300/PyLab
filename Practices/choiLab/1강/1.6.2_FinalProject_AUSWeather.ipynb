## 계절성 특징 추출
이제 `날짜` 열을 살펴보겠습니다. 날씨 패턴은 계절성을 띠며, 예를 들어 겨울과 여름의 예측 가능성이 다를 것으로 예상됩니다.
`연도`에도 약간의 변동이 있을 수 있지만, 지금은 고려하지 않겠습니다.
`날짜`에서 `계절` 특징을 추출하고, `날짜` 열 자체는 계절보다 정보량이 적을 가능성이 높으므로 나중에 제거하겠습니다.
이를 위한 간단한 방법은 주어진 월에 계절을 할당하는 함수를 정의한 다음, 해당 함수를 사용하여 `날짜` 열을 변환하는 것입니다.

### Create a function to map dates to seasons


````python
def date_to_season(date):
    month = date.month
    if (month == 12) or (month == 1) or (month == 2):
        return 'Summer'
    elif (month == 3) or (month == 4) or (month == 5):
        return 'Autumn'
    elif (month == 6) or (month == 7) or (month == 8):
        return 'Winter'
    elif (month == 9) or (month == 10) or (month == 11):
        return 'Spring'
````

## 연습 7. 트랜스포머들을 하나의 전처리 열 트랜스포머로 결합하세요
Complete the followng code:  
```python
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, ...),
        ('cat', categorical_transformer, ...)
    ]
)
```


````python
# Write your response.
preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)

preprocessor
````

## Exercise 8. Create a pipeline by combining the preprocessing with a Random Forest classifier
## 연습문제 8. 전처리 단계와 랜덤 포레스트 분류기를 결합하여 파이프라인을 구축하세요.
Complete the following code:
```python
pipeline = Pipeline(steps=[
    ('preprocessor', ...),
    ('...', RandomForestClassifier(random_state=42))
])
```


````python
# Write your response.
pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier(random_state=42))
])
pipeline
````

````python
# Combine numeric and categorical feature names
feature_names = numeric_features + list(grid_search.best_estimator_['preprocessor']
                                        .named_transformers_['cat']
                                        .named_steps['onehot']
                                        .get_feature_names_out(categorical_features))

feature_importances = grid_search.best_estimator_['classifier'].feature_importances_

importance_df = pd.DataFrame({'Feature': feature_names,
                              'Importance': feature_importances
                             }).sort_values(by='Importance', ascending=False)

N = 20  # Change this number to display more or fewer features
top_features = importance_df.head(N)

# Plotting
plt.figure(figsize=(10, 6))
plt.barh(top_features['Feature'], top_features['Importance'], color='skyblue')
plt.gca().invert_yaxis()  # Invert y-axis to show the most important feature on top
plt.title(f'Top {N} Most Important Features in predicting whether it will rain today')
plt.xlabel('Importance Score')
plt.show()
````
